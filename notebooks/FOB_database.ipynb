```python
import os, sys
import pandas as pd
from filelock import FileLock
```

```python
cwd = os.getcwd()
raw_path = os.path.join(os.path.dirname(cwd),'data','raw','FOB')
zip_files = [f for f in os.listdir(raw_path) if os.path.splitext(f)[1] == '.zip']
display(zip_files)
```

```python
if 'FOB_DB.csv' in os.listdir(cwd):
    display('ok')
    
else:
    sys.exit('FOB_DB.csv does not exist.')
    
df = pd.read_csv(os.path.join(cwd, 'FOB_DB.csv'), index_col=0)
display(df)
```

```python
if df.empty:
    ls = pd.read_csv(os.path.join(raw_path, os.path.splitext(zip_files[0])[0]), usecols=['isin'])['isin'].unique().tolist()
    df['isin'] = ls
    df[['file','state','allocate']] = zip_files[0],'Pending',None
    display(df)
```

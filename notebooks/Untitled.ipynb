```python
import pandas as pd
from fastparquet import write, ParquetFile

# Étape 1 : Création du fichier Parquet initial compressé
# Créer un DataFrame initial
df_initial = pd.DataFrame({
    'Colonne1': [1, 2, 3],
    'Colonne2': ['A', 'B', 'C']
})

# Écrire le DataFrame initial dans un fichier Parquet compressé avec gzip
write('mon_fichier_compressé.parquet', df_initial, compression='GZIP')

# Étape 2 : Ajout de nouvelles données au fichier Parquet compressé
# Créer un nouveau DataFrame avec des données supplémentaires
nouveau_df = pd.DataFrame({
    'Colonne1': [4, 5, 6],
    'Colonne2': ['D', 'E', 'F']
})

# Ajouter les nouvelles données au fichier Parquet compressé en conservant la compression
write('mon_fichier_compressé.parquet', nouveau_df, append=True, compression='GZIP')

# Vérification : Lire le fichier Parquet pour confirmer l'ajout
parquet_file = ParquetFile('mon_fichier_compressé.parquet')
df_complet = parquet_file.to_pandas()
print(df_complet)
```
